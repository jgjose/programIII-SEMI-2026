```python
# Diccionario con las unidades de superficie de El Salvador y sus factores base en metros cuadrados
unidades = {
    1: ("Pie Cuadrado", 0.092903),
    2: ("Vara Cuadrada", 0.698742),
    3: ("Yarda Cuadrada", 0.836127),
    4: ("Metro Cuadrado", 1.0),
    5: ("Tareas", 436.7135),
    6: ("Manzana", 6987.416),
    7: ("Hectárea", 10000.0)
}

def mostrar_menu():
    print("\n--- MENÚ DE UNIDADES DE SUPERFICIE ---")
    for clave, datos in unidades.items():
        print(f"[{clave}] {datos[0]}")

def convertir():
    print("=== CONVERSOR DE ÁREAS (EL SALVADOR) ===")
    mostrar_menu()
    try:
        opcion_origen = int(input("\nElige el número de la unidad de ORIGEN: "))
        if opcion_origen not in unidades:
            print("[!] Opción de origen no válida.")
            return
        mostrar_menu()
        opcion_destino = int(input("\nElige el número de la unidad de DESTINO: "))
        if opcion_destino not in unidades:
            print("[!] Opción de destino no válida.")
            return

        cantidad = float(input(f"Ingresa la cantidad de [{unidades[opcion_origen][0]}]: "))

        if cantidad < 0:
            print("[!] La cantidad no puede ser negativa.")
            return
        valor_en_metros = cantidad * unidades[opcion_origen][1]
        resultado = valor_en_metros / unidades[opcion_destino][1]
        print("\n" + "="*40)
        print(f" [RESULTADO]: {cantidad} {unidades[opcion_origen][0]} ")
        print(f" Equivalen a: {resultado:,.4f} {unidades[opcion_destino][0]}")
        print("="*40)

    except ValueError:
        print("[!] Error: Debes ingresar números enteros para las opciones y valores numéricos válidos para la cantidad.")
if __name__ == "__main__":
    convertir()
```

```python
class SistemaEstadistico:
    def __init__(self, datos):

        self.datos = sorted([float(x) for x in datos])
        self.n = len(self.datos)

    def media_aritmetica(self):
        return sum(self.datos) / self.n

    def mediana(self):
        medio = self.n // 2
        if self.n % 2 == 0:
            return (self.datos[medio - 1] + self.datos[medio]) / 2
        else:
            return self.datos[medio]

    def moda(self):
        frecuencias = self.frecuencia()
        max_frecuencia = max(frecuencias.values())

        modas = [k for k, v in frecuencias.items() if v == max_frecuencia]
        return modas

    def varianza(self):
        media = self.media_aritmetica()
        suma_cuadrados = sum((x - media) ** 2 for x in self.datos)
        return suma_cuadrados / self.n

    def desviacion_estandar(self):
        return self.varianza() ** 0.5

    def desviacion_tipica(self):

        return self.desviacion_estandar()

    def rango(self):
        return self.datos[-1] - self.datos[0]

    def frecuencia(self):
        conteo = {}
        for x in self.datos:

            val = int(x) if x.is_integer() else x
            conteo[val] = conteo.get(val, 0) + 1
        return conteo



muestra_datos = [18, 20, 19, 21, 20, 22, 19, 20, 25, 21]
est = SistemaEstadistico(muestra_datos)

print("--- RESULTADOS ESTADÍSTICOS ---")
print(f"Datos analizados: {est.datos}")
print(f"Media aritmética: {est.media_aritmetica():.2f}")
print(f"Mediana: {est.mediana()}")
print(f"Moda: {est.moda()}")
print(f"Varianza: {est.varianza():.2f}")
print(f"Desviación estándar: {est.desviacion_estandar():.2f}")
print(f"Desviación típica: {est.desviacion_tipica():.2f}")
print(f"Rango: {est.rango()}")
print(f"Frecuencias: {est.frecuencia()}")
```
