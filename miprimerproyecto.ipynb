```python
# --- CLASE ESTADISTICA ---
class Estadistica:
    def media(self, datos):
        datos = list(map(float, datos))
        return sum(datos) / len(datos)

    def varianza(self, datos):
        datos = list(map(float, datos))
        media = self.media(datos)
        return sum((x - media) ** 2 for x in datos) / len(datos)

    def desviacion(self, datos):
        return self.varianza(datos) ** 0.5

    def media_armonica(self, datos):
        datos = list(map(float, datos))
        return len(datos) / sum(1/x for x in datos)


# --- CLASE OPERACIONES ---
class Operaciones:
    def frecuencia(self, lista):
        freq = {}
        for elemento in lista:
            freq[elemento] = freq.get(elemento, 0) + 1
        return freq


# --- CLASE PALINDROMO ---
class Palindromo:
    def __init__(self, palabra):
        self.palabra = palabra.lower()

    def es_palindromo(self):
        return self.palabra == self.palabra[::-1]


# --- MENÚ PRINCIPAL ---
def menu():
    print("\n===== MENÚ DE TAREAS =====")
    print("1. Estadística (media, varianza, desviación, media armónica)")
    print("2. Calcular frecuencia de elementos")
    print("3. Verificar si una palabra es palíndromo")
    print("4. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        serie = input("Serie de números separados por coma: ").split(",")
        obj = Estadistica()
        print("➡ Media:", obj.media(serie))
        print("➡ Varianza:", obj.varianza(serie))
        print("➡ Desviación estándar:", obj.desviacion(serie))
        print("➡ Media armónica:", obj.media_armonica(serie))

    elif opcion == "2":
        lista = input("Ingrese elementos separados por espacio: ").split()
        op = Operaciones()
        print("➡ Frecuencia:", op.frecuencia(lista))

    elif opcion == "3":
        palabra = input("Ingrese una palabra: ")
        obj = Palindromo(palabra)
        if obj.es_palindromo():
            print(f"➡ '{palabra}' es un palíndromo ✅")
        else:
            print(f"➡ '{palabra}' no es un palíndromo ❌")

    elif opcion == "4":
        print("Programa finalizado.")
        return False

    else:
        print("Opción inválida. Intente de nuevo.")

    return True


# --- EJECUCIÓN ---
while True:
    if not menu():
        break
```

```python
class SistemaEstadistico:
    def __init__(self, datos):

        self.datos = sorted([float(x) for x in datos])
        self.n = len(self.datos)

    def media_aritmetica(self):
        return sum(self.datos) / self.n

    def mediana(self):
        medio = self.n // 2
        if self.n % 2 == 0:
            return (self.datos[medio - 1] + self.datos[medio]) / 2
        else:
            return self.datos[medio]

    def moda(self):
        frecuencias = self.frecuencia()
        max_frecuencia = max(frecuencias.values())
        # Si todos se repiten igual, podría no haber moda única
        modas = [k for k, v in frecuencias.items() if v == max_frecuencia]
        return modas

    def varianza(self):
        media = self.media_aritmetica()
        suma_cuadrados = sum((x - media) ** 2 for x in self.datos)
        return suma_cuadrados / self.n

    def desviacion_estandar(self):
        return self.varianza() ** 0.5

    def desviacion_tipica(self):

        return self.desviacion_estandar()

    def rango(self):
        return self.datos[-1] - self.datos[0]

    def frecuencia(self):
        conteo = {}
        for x in self.datos:

            val = int(x) if x.is_integer() else x
            conteo[val] = conteo.get(val, 0) + 1
        return conteo



muestra_datos = [18, 20, 19, 21, 20, 22, 19, 20, 25, 21]
est = SistemaEstadistico(muestra_datos)

print("--- RESULTADOS ESTADÍSTICOS ---")
print(f"Datos analizados: {est.datos}")
print(f"Media aritmética: {est.media_aritmetica():.2f}")
print(f"Mediana: {est.mediana()}")
print(f"Moda: {est.moda()}")
print(f"Varianza: {est.varianza():.2f}")
print(f"Desviación estándar: {est.desviacion_estandar():.2f}")
print(f"Desviación típica: {est.desviacion_tipica():.2f}")
print(f"Rango: {est.rango()}")
print(f"Frecuencias: {est.frecuencia()}")
```
