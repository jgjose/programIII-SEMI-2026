Ejercicio encontrar la NF de cada materia, haciendo uso de funciones

```python
continuar = True

while continuar:
    numero = int(input("Introduce un número entero: "))
    if numero % 2 == 0:
        print(f"El número {numero} es PAR.")
    else:
        print(f"El número {numero} es IMPAR.")
    if numero <= 1:
        print(f"El número {numero} NO es primo (deben ser mayores a 1).")
    else:
        es_primo = True
        for i in range(2, numero):
            if numero % i == 0:
                es_primo = False
                break
        if es_primo:
            print(f"El número {numero} es PRIMO.")
        else:
            print(f"El número {numero} NO es primo.")
    print("-" * 30)
    respuesta = input("¿Deseas evaluar otro número? (Presiona 's' para seguir, o cualquier otra tecla para salir): ")

    if respuesta.lower() != 's':
        continuar = False
        print("¡Hasta luego!")
```

```python
# --- CLASE ESTADISTICA ---
class Estadistica:
    def media(self, datos):
        datos = list(map(float, datos))
        return sum(datos) / len(datos)

    def varianza(self, datos):
        datos = list(map(float, datos))
        media = self.media(datos)
        return sum((x - media) ** 2 for x in datos) / len(datos)

    def desviacion(self, datos):
        return self.varianza(datos) ** 0.5

    def media_armonica(self, datos):
        datos = list(map(float, datos))
        return len(datos) / sum(1/x for x in datos)


# --- CLASE OPERACIONES ---
class Operaciones:
    def frecuencia(self, lista):
        freq = {}
        for elemento in lista:
            freq[elemento] = freq.get(elemento, 0) + 1
        return freq


# --- CLASE PALINDROMO ---
class Palindromo:
    def __init__(self, palabra):
        self.palabra = palabra.lower()

    def es_palindromo(self):
        return self.palabra == self.palabra[::-1]


# --- MENÚ PRINCIPAL ---
def menu():
    print("\n===== MENÚ DE TAREAS =====")
    print("1. Estadística (media, varianza, desviación, media armónica)")
    print("2. Calcular frecuencia de elementos")
    print("3. Verificar si una palabra es palíndromo")
    print("4. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        serie = input("Serie de números separados por coma: ").split(",")
        obj = Estadistica()
        print("➡ Media:", obj.media(serie))
        print("➡ Varianza:", obj.varianza(serie))
        print("➡ Desviación estándar:", obj.desviacion(serie))
        print("➡ Media armónica:", obj.media_armonica(serie))

    elif opcion == "2":
        lista = input("Ingrese elementos separados por espacio: ").split()
        op = Operaciones()
        print("➡ Frecuencia:", op.frecuencia(lista))

    elif opcion == "3":
        palabra = input("Ingrese una palabra: ")
        obj = Palindromo(palabra)
        if obj.es_palindromo():
            print(f"➡ '{palabra}' es un palíndromo ✅")
        else:
            print(f"➡ '{palabra}' no es un palíndromo ❌")

    elif opcion == "4":
        print("Programa finalizado.")
        return False

    else:
        print("Opción inválida. Intente de nuevo.")

    return True


# --- EJECUCIÓN ---
while True:
    if not menu():
        break
```
